```python
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
documents = [
    "Artificial Intelligence (AI) is a field focused on building intelligent systems.",
    "Machine Learning allows computers to learn from data without being explicitly programmed.",
    "Deep Learning uses neural networks with multiple layers.",
    "Natural Language Processing helps computers understand human language.",
    "Transformers are powerful models used in modern NLP like ChatGPT.",
    "FAISS is a library used for fast similarity search in vector databases."
]

print("Documents created:", len(documents))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=10
)

docs = splitter.create_documents(documents)

print("Chunks created:", len(docs))
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embeddings model loaded")
```

```python
from langchain_community.vectorstores import FAISS

db = FAISS.from_documents(docs, embeddings)

print("FAISS vector database created")
```

```python
query = "What is deep learning?"

results = db.similarity_search(query, k=2)

for i, r in enumerate(results):
    print(f"\nResult {i+1}:")
    print(r.page_content)
```

```python
def chat(query):
    results = db.similarity_search(query, k=2)

    context = "\n".join([r.page_content for r in results])

    response = f"""
Question: {query}

Relevant Context:
{context}

Answer:
Based on the retrieved context, this topic is related to AI/ML concepts explained in the dataset.
"""
    return response
```

```python
print(chat("What is machine learning?"))
```
